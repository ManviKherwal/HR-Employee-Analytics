```python
# ===========================================
# Feature 2 : Income Category
# ===========================================

def income_category(income):
    if income <= 3000:
        return "Low Income"
    elif income <= 8500:
        return "Medium Income"
    else:
        return "High Income"

df["Income_Category"] = df["MonthlyIncome"].apply(income_category)
```

```python
df[["MonthlyIncome","Income_Category"]].head()
```

```python
df["MonthlyIncome"].describe()
```

```python
df["Income_Category"].value_counts()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Income_Category",color="#4F81BD")
plt.title("Employee Distribution by Income Category ")
plt.show()
```

# Business Insight
- Employees are grouped into income categories to simplify salary analysis.
- Income groups help compare compensation and attrition trends.

```python
# ===========================================
# Feature 3 : Experience Level
# ===========================================

def experience_level(years):
    if years <= 5:
        return "Fresher"
    elif years <= 15:
        return "Mid level"
    else:
        return "Experienced"

df["Experience_Level"] = df["TotalWorkingYears"].apply(experience_level)
```

```python
df[["TotalWorkingYears", "Experience_Level"]].head()
```

```python
df["Experience_Level"].value_counts()
```

```python
plt.figure(figsize=(7,4))
sns.countplot(data=df, x="Experience_Level",color="#4F81BD")
plt.title("Employee Distribution by Experience Level")
plt.show()
```

# Business Insight:
- Employees are classified based on their work experience.
- Experience levels support workforce planning and retention analysis.

```python
# ===========================================
# Feature 4 : Tenure Group
# ===========================================

def tenure_group(years):
    if years <= 2:
        return "New Employee"
    elif years <= 7:
        return "Mid Tenure"
    else:
        return "Long Tenure"

df["Tenure_Group"] = df["YearsAtCompany"].apply(tenure_group)

df[["YearsAtCompany", "Tenure_Group"]].head()
```

```python
df["Tenure_Group"].value_counts()
```

```python
plt.figure(figsize=(7,4))

sns.countplot(data=df, x="Tenure_Group",color="#4F81BD")

plt.title("Employee Distribution by Tenure Group")

plt.show()
```

# Business Insight
- Employees are grouped based on their tenure in the company.
- Tenure groups help analyze employee retention patterns.
